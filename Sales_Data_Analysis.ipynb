```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Product": [
        "AirPods Pro 2", "Smart Watch T900", "Anker Cable",
        "AirPods Pro 4", "Smart Watch T1000", "Anker Cable",
        "AirPods Pro 2", "Smart Watch T2000", "AirPods Pro 4",
        "Anker Cable", "Smart Watch T900", "AirPods Pro 2"
    ],

    "Category": [
        "AirPods", "Smart Watches", "Cables",
        "AirPods", "Smart Watches", "Cables",
        "AirPods", "Smart Watches", "AirPods",
        "Cables", "Smart Watches", "AirPods"
    ],

    "Price": [
        299, 249, 35, 349, 279, 35,
        299, 319, 349, 35, 249, 299
    ],

    "Quantity": [
        3, 2, 5, 2, 3, 8,
        4, 2, 3, 10, 5, 6
    ],

    "Month": [
        "January", "January", "January",
        "February", "February", "February",
        "March", "March", "March",
        "April", "April", "April"
    ]
}

df = pd.DataFrame(data)

df["Total_Sales"] = df["Price"] * df["Quantity"]

df
```

```python
print("Total Sales:", df["Total_Sales"].sum())
print("Total Quantity:", df["Quantity"].sum())

print("\nTop Products:")
print(
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
category_sales.plot(
    kind="bar",
    figsize=(8, 5),
    title="Sales by Category"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales = (
    df.groupby("Month")["Total_Sales"]
    .sum()
)

print(monthly_sales)
```

```python
monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(8, 5),
    title="Monthly Sales"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
top_products = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

top_products
```

```python
top_products.head(5).plot(
    kind="bar",
    figsize=(9, 5),
    title="Top 5 Products by Sales"
)

plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
total_sales = df["Total_Sales"].sum()
total_quantity = df["Quantity"].sum()
best_product = top_products.index[0]
best_category = category_sales.index[0]

print("===== SALES ANALYSIS SUMMARY =====")
print("Total Sales:", total_sales)
print("Total Quantity Sold:", total_quantity)
print("Best Product:", best_product)
print("Best Category:", best_category)
```

```python
print("===== BUSINESS INSIGHTS =====")

print(f"1. Total sales: {total_sales}")

print(f"2. Total products sold: {total_quantity}")

print(f"3. Best-selling product: {best_product}")

print(f"4. Best-selling category: {best_category}")

best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print(f"5. Best-selling month: {best_month}")
print(f"   Month's sales: {best_month_sales}")
```
